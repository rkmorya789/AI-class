```python
city_name=["mohali","mansa","chd","mansa","hanumangard",]
price=[200,300,400,500,600]
unique_city=[]
citywise_sale=[]
for i in city_name:
  if i not in  unique_city:
    unique_city.append(i)
    citywise_sale.append(0)
print(citywise_sale)
for i in range(len(city_name)):
  index=unique_city.index(city_name[i])
  citywise_sale[index]=citywise_sale[index]+price[i]
print(citywise_sale)


```

```python
import matplotlib.pyplot as plt
plt.plot(unique_city,citywise_sale)
plt.xlabel(unique_city)
plt.ylabel(citywise_sale)
plt.show()
```
