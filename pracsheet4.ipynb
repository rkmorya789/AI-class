```python
name=[]
price=[]
discount=[]
total=0
pay_bill=0
n=int(input("enter the list item number "))
for i in range(n):
  names=input("enter the items name ")
  prices=int(input("enter the item price "))
  discounts=int(input("enter the discount apply for precentage "))
  name.append(names)
  price.append(prices)
  discount.append(discounts)
  total=total+prices
  apply_discount=prices*discounts/100
  discounted_bill=prices-apply_discount
  pay_bill=pay_bill+discounted_bill
print("total bill for all amount without discount",total)
print("your total bill you pay",pay_bill)
```
