```python
num=int(input("enter the num "))
rev=0
while num>0:
  digit=num%10
  rev=rev*10+digit
  num=num//10
print(rev)
```

```python
num=int(input("enter the num "))
rev=0
tem=num
while num>0:
  digit=num%10
  rev=rev*10+digit
  num=num//10
print(rev)
if rev==tem:
    print("palindrom")
else:
    print("not palindrom")
```
